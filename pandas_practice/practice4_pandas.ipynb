```python
df1 = pd.DataFrame({
    'Name':['A','B'],
    'Marks':[80,90]
})

df2 = pd.DataFrame({
    'Name':['C','D'],
    'Marks':[70,60]
})
```

```python
students = pd.DataFrame({
    'id':[1,2,3],
    'Name':['Ram','Shyam','Mohan']
})

marks = pd.DataFrame({
    'id':[1,2,3],
    'Marks':[90,85,88]
})
```

```python
students
```

```python
marks
```

```python
pd.merge(students,marks,on='id')
```

```python
employees = pd.DataFrame({
    'id':[1,2,3,4],
    'Name':['A','B','C','D']
})

salary = pd.DataFrame({
    'id':[1,2],
    'Salary':[50000,60000]
})
```

```python
employees
```

```python
salary
```

```python
pd.merge(employees,salary,how='left')
```

```python
df = pd.DataFrame({
    'Department':['IT','HR','IT','HR','Sales'],
    'Salary':[50000,40000,60000,45000,55000]
})
```

```python
df
```

```python
a = df.groupby(['Department'])
for i , v in a:
    print(i)
    print(v)
```

```python
df.groupby('Department')['Salary'].agg(['mean','max','min'])
```
